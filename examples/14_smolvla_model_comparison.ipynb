# Our fine-tuned SmolVLA against the two pretrained LIBERO checkpoints

Four policies on the **same scene** (LIBERO's table and robot, cleared, with the bowl and plate at their real, scaled sizes: the `scaled` condition) and the **same layouts**:

| Policy | Weights |
|---|---|
| ours, `smolvla_base` init | `lerobot/smolvla_base` fine-tuned on our 100 generated episodes (action expert only); the best of its checkpoints |
| ours, `smolvla_libero` init | `lerobot/smolvla_libero` fine-tuned on the same data with the same settings; the best of its checkpoints |
| `HuggingFaceVLA/smolvla_libero` | pretrained, used as is |
| `lerobot/smolvla_libero` | pretrained, used as is |

Modules: `hurobo.model_comparison`, `hurobo.checkpoint_eval`, `hurobo.policy_rollout`. Real data only: 12 real clips (7 passing the QC gates, 5 with the release-distance gate waived) turned into 100 robot episodes.

**How the best checkpoint is chosen.** By closed-loop success on layouts 0 to 9 (ties: lower open-loop error). It is then compared on **held-out layouts 100 and up**, so it is not judged on the episodes it was picked on. The two pretrained policies are run on the same held-out layouts. All policies use the same episode limit and replan interval. The sample is small, so read the intervals, not just the bars.

```python
from hurobo.checkpoint_eval import checkpoint_table
from hurobo.model_comparison import ModelComparison, fine_tuned_contender
from hurobo.policy_rollout import KNOWN_CHECKPOINTS
from hurobo.real_setup import DEFAULT_MEASUREMENTS, RealSetup

FINAL_STEP = 3000
RUNS = {
    "ours, smolvla_base init": (ROOT / "outputs/e1_smolvla_ckpts/run", ROOT / "results/smolvla_ckpts"),
    "ours, smolvla_libero init": (
        ROOT / "outputs/e1_smolvla_libero_init/run",
        ROOT / "results/smolvla_libero_init_ckpts",
    ),
}
contenders = []
for label, (run_dir, results_dir) in RUNS.items():
    try:
        contenders.append(fine_tuned_contender(label, run_dir, results_dir, FINAL_STEP))
    except ValueError as reason:
        print(f"not entered: {label}: {reason}")
contenders += [KNOWN_CHECKPOINTS["hfvla"], KNOWN_CHECKPOINTS["lerobot"]]
print([spec.name for spec in contenders])
```

## How each training run progressed

Closed-loop success per checkpoint step (layouts 0 to 9, 10 episodes each) with the 95 % Wilson interval. This is what the best checkpoint is picked from.

```python
from hurobo.utils.comparison_view import progress_figure

progress_figure({label: checkpoint_table(results) for label, (_, results) in RUNS.items()}).show()
```

## Closed loop on held-out layouts, scaled scene

```python
from hurobo.utils.comparison_view import success_figure

comparison = ModelComparison(RealSetup.load(DEFAULT_MEASUREMENTS), ROOT / "results/smolvla_compare")
summary = comparison.run(contenders, episodes=20)
summary[
    [
        "checkpoint",
        "episodes",
        "successes",
        "success_rate",
        "ci_low",
        "ci_high",
        "lifted",
        "mean_steps_success",
    ]
]
```

```python
success_figure(summary).show()
```

## Open loop against the generated data

The error of the predicted action chunk against the demonstrations, on the same sampled frames for every policy. The fine-tuned policies trained on this dataset; the pretrained ones did not, so their error here says how far they are from our demonstrations, not how good they are.

```python
comparison.open_loop(contenders, ROOT / "data/e1/generated_clips").round(3)
```
